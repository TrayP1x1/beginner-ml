### Python Pro Tip: See The Real Value With `repr`

If you are not sure what a value really looks like, inspect both its type and its representation.

```python
value = "hello\n"
print(type(value))
print(f"{value!r}")
print(f"{value=}")
```

- `!r` calls `repr(value)`, which shows extra detail such as quotes and escaped newline characters
- `f"{value=}"` prints both the variable name and the value, which is great for quick debugging
- `type(value)` confirms whether you are working with a `str`, `int`, `float`, and so on


````python
name = "Tray"
age = 21
wants_student_dev_ai_role = True
hours_practiced_today = 1.5

print(type(name), name)
print(type(age), age)
print(type(wants_student_dev_ai_role), wants_student_dev_ai_role)
print(type(hours_practiced_today), hours_practiced_today)

number_as_text = "42"
number_as_int = int(number_as_text)
number_as_float = float(number_as_text)

print(number_as_text, type(number_as_text))
print(number_as_int, type(number_as_int))
print(number_as_float, type(number_as_float))
````

## Day 2 - Conditionals, Loops, and Lists

### Core idea

Real data is repetitive.

- A `list` stores many items in order.
- An `if` statement lets your program make a decision.

### Why this matters for AI

### Python Pro Tip: Loop Clearly

- use `enumerate(scores, start=1)` when you need both the index and the value
- build new lists with `append(...)` instead of manually tracking positions
- compare your loop result with a built-in like `sum(scores)` to sanity-check your logic


````python
scores = [4, 7, 10, 13, 18]

even_scores = []
total = 0

for score in scores:
    total += score
    if score % 2 == 0:
        even_scores.append(score)

print("scores:", scores)
print("even scores:", even_scores)
print("total:", total)
print("average:", total / len(scores))
````

### Exercise Pattern: Write Functions First

A strong beginner workflow is: write one small function, test it on a tiny example, then build `main()` to connect the steps.

```python
def load_scores(path):
    ...

def summarize_scores(rows):
    ...

def main():
    rows = load_scores("data/tiny_scores.csv")
    summarize_scores(rows)

if __name__ == "__main__":
    main()
```

This makes your code easier to test, reuse, and debug than placing every step at the top level. In notebook demos, top-level code is fine for quick learning. In exercise files, prefer this structure whenever the task has multiple steps.


````python
def add(a, b):
    return a + b

def average(numbers):
    return sum(numbers) / len(numbers)

def is_passing(score, passing_grade=5.5):
    return score >= passing_grade

print(add(3, 8))
print(average([6, 7, 8, 9]))
print(is_passing(5.0))
print(is_passing(7.2))
````

## Day 4 - Dictionaries, Tuples, and Reading CSV Files With The Standard Library

### Core idea

Before using `pandas`, it is worth understanding what a CSV file really is.
A CSV file is plain text where each row contains comma-separated values.

The built-in `csv` module reads rows one by one.
This is slower and less convenient than `pandas`, but it teaches the structure clearly.

- A `tuple` is like a list, but usually used for fixed records.
- A `dict` maps keys to values.

### Why this matters for AI

When you load a CSV into `pandas`, you are still working with rows, columns, and values.
Understanding the simple version first removes a lot of confusion.

### Daily exercises

### Bootcamp exercise folder

### Python Pro Tip: Let `DictReader` Do The Labeling

- `csv.DictReader(...)` gives each column a name, which is easier to reason about than raw positions
- use `row.get("score")` when a key might be missing
- print one full row with `print(row)` before writing more logic so you know the exact structure


````python
Path("data").mkdir(exist_ok=True)

tiny_csv_path = Path("data/tiny_scores.csv")

with tiny_csv_path.open("w", newline="") as file:
    writer = csv.writer(file)
    writer.writerow(["name", "python_score", "numpy_score"])
    writer.writerow(["Alice", 8, 7])
    writer.writerow(["Bob", 6, 5])
    writer.writerow(["Cara", 9, 8])

rows = []
with tiny_csv_path.open(newline="") as file:
    reader = csv.reader(file)
    header = next(reader)
    for row in reader:
        rows.append(row)

print("header:", header)
print("rows:", rows)

average_python = sum(int(row[1]) for row in rows) / len(rows)
print("average python score:", average_python)
````

## Day 8 - Creating and Loading a Practice CSV With Pandas

We will create one synthetic dataset that we can use for the rest of the bootcamp.
It will contain realistic tabular features and two targets:

- `test_score`: a number for regression
- `passed_test`: a 0/1 label for classification

This is useful because a real coding test may ask for either regression or classification.


### Pandas Pro Tip: Inspect The Table Before Touching It

- start with `df.head()` so you can see real rows immediately
- use `df.columns.tolist()` to confirm the exact column names
- use `df.shape` early so you always know how much data you are working with


````python
rng = np.random.default_rng(42)
n = 240

hours_python = rng.normal(5.5, 1.5, n).clip(1, 10)
hours_numpy = rng.normal(4.5, 1.2, n).clip(0.5, 9)
projects_completed = rng.integers(0, 9, n)
attendance_rate = rng.normal(0.86, 0.08, n).clip(0.55, 1.0)
practice_tests = rng.integers(0, 7, n)
sleep_hours = rng.normal(7.1, 0.9, n).clip(4.5, 9.5)

raw_score = (
    hours_python * 5.5
    + hours_numpy * 4.5
    + projects_completed * 2.5
    + attendance_rate * 18
    + practice_tests * 2.8
    + sleep_hours * 1.3
    + rng.normal(0, 4, n)
)

test_score = raw_score.clip(35, 100)
passed_test = (test_score >= 70).astype(int)

practice_df = pd.DataFrame({
    "hours_python": hours_python.round(2),
    "hours_numpy": hours_numpy.round(2),
    "projects_completed": projects_completed,
    "attendance_rate": attendance_rate.round(2),
    "practice_tests": practice_tests,
    "sleep_hours": sleep_hours.round(2),
    "test_score": test_score.round(1),
    "passed_test": passed_test,
})

# Add a few missing values on purpose so we can practice cleaning.
practice_df.loc[[5, 21, 120], "hours_numpy"] = np.nan
practice_df.loc[[17, 88], "sleep_hours"] = np.nan

Path("data").mkdir(exist_ok=True)

csv_path = Path("data/student_dev_ai_practice.csv")
practice_df.to_csv(csv_path, index=False)

print(f"Saved practice dataset to: {csv_path.resolve()}")
practice_df.head()
````

## Day 10 - Data Cleaning and Missing Values

### Core idea

Real CSV files are messy.
Some values are missing, some columns use the wrong type, and sometimes rows are duplicated or malformed.

You should always check:

- `df.isna().sum()`
- column types with `df.info()`
- suspicious values like negatives where they should not exist

### Common choices for missing values

- drop the row
- fill with the mean
- fill with the median
- fill with a constant

There is no single perfect answer. What matters is that you can explain your choice.

### Daily exercises

- Count missing values per column.
- Fill numeric missing values with the column mean.
- Verify there are no missing values left.

### Bootcamp exercise folder

- `Week2/day10_clean_missing_values/README.md`

### Cleaning Tip: Keep A Raw Copy

- keep the original data untouched and clean a copy with `cleaned_df = df.copy()`
- compare `df.isna().sum()` before and after cleaning so you can prove what changed
- make one cleaning decision at a time so mistakes are easier to trace


````python
print("Missing values before cleaning:")
display(df.isna().sum())

cleaned_df = df.copy()
numeric_means = cleaned_df.mean(numeric_only=True)
cleaned_df = cleaned_df.fillna(numeric_means)

print("Missing values after cleaning:")
display(cleaned_df.isna().sum())
display(cleaned_df.head())
````

## Day 13 - Train, Validation, Test Split and Normalization

### Core idea

Never train and evaluate on the exact same rows if you want an honest result.

Common split idea:

- training set: used to fit the model
- validation set: used to tune decisions
- test set: used once at the end for final evaluation

### Why normalization matters

Neural networks train better when numeric features are on similar scales.
A simple standardization is:

`normalized = (value - mean) / std`

Important rule: compute the mean and std on the training set only, then apply those values to validation and test sets.

### Daily exercises

- Split the dataset into train, validation, and test.
- Normalize the feature columns.
- Verify the training features now have roughly mean `0`.


### Split Tip: Learn Statistics From Train Only

- compute means and standard deviations on the training split only
- reuse those same values for validation and test data
- save `means` and `stds` in variables so the normalization logic stays explicit and reusable


````python
shuffled_df = engineered_df.sample(frac=1, random_state=42).reset_index(drop=True)

n_total = len(shuffled_df)
n_train = int(n_total * 0.7)
n_val = int(n_total * 0.15)

train_df = shuffled_df.iloc[:n_train].copy()
val_df = shuffled_df.iloc[n_train:n_train + n_val].copy()
test_df = shuffled_df.iloc[n_train + n_val:].copy()

feature_columns = [
    "hours_python",
    "hours_numpy",
    "projects_completed",
    "attendance_rate",
    "practice_tests",
    "sleep_hours",
    "total_study_hours",
    "projects_per_test",
]

train_means = train_df[feature_columns].mean()
train_stds = train_df[feature_columns].std().replace(0, 1)

X_train_df = (train_df[feature_columns] - train_means) / train_stds
X_val_df = (val_df[feature_columns] - train_means) / train_stds
X_test_df = (test_df[feature_columns] - train_means) / train_stds

print("train shape:", X_train_df.shape)
print("val shape:", X_val_df.shape)
print("test shape:", X_test_df.shape)
print("approx training means after normalization:")
display(X_train_df.mean().round(2))
````

## Day 15 - Tensor Basics and The Shape Mindset

### Core idea

A tensor is PyTorch's array object.
If you understand NumPy arrays, tensors will feel familiar.

Key ideas:

- tabular input usually has shape `(n_samples, n_features)`
- a regression target is often shape `(n_samples, 1)`
- a binary classification target is also often shape `(n_samples, 1)`

Beginners often get blocked by shape mistakes, not by hard math.
So always print shapes.

### Daily exercises

### Tensor Tip: Shapes Explain Most Errors

- print `tensor.shape` whenever a model step feels confusing
- print `tensor.dtype` if a PyTorch operation complains about types
- remember that a single target column often needs shape `(n, 1)` rather than `(n,)`


````python
X_train_tensor = torch.tensor(X_train_df.values, dtype=torch.float32)
X_val_tensor = torch.tensor(X_val_df.values, dtype=torch.float32)
X_test_tensor = torch.tensor(X_test_df.values, dtype=torch.float32)

y_train_reg = torch.tensor(train_df[["test_score"]].values, dtype=torch.float32)
y_val_reg = torch.tensor(val_df[["test_score"]].values, dtype=torch.float32)
y_test_reg = torch.tensor(test_df[["test_score"]].values, dtype=torch.float32)

y_train_clf = torch.tensor(train_df[["passed_test"]].values, dtype=torch.float32)
y_val_clf = torch.tensor(val_df[["passed_test"]].values, dtype=torch.float32)
y_test_clf = torch.tensor(test_df[["passed_test"]].values, dtype=torch.float32)

print("X_train_tensor shape:", X_train_tensor.shape)
print("y_train_reg shape:", y_train_reg.shape)
print("y_train_clf shape:", y_train_clf.shape)
````

## Day 16 - Regression In PyTorch

### Problem type

Regression means predicting a number.
In our dataset, `test_score` is numeric, so it is a regression target.

### Model structure

We will use a very small neural network:

- input layer with one node per feature
- hidden layer with `ReLU`
- output layer with one value

### Loss choice

For regression, a standard choice is `nn.MSELoss()`.
MSE means mean squared error.

### Daily exercises

- Train the model.
- Track training and validation loss.
- Plot the loss curves.
- Compute the mean absolute error on the test set.

### Bootcamp exercise folder

- `Week3/day16_pytorch_regression/README.md`
- Best real regression dataset: `data/kaggle/housing_prices.csv`
- Useful bonus inside this pack: `Bonus/bonus_least_squares_regression.md`


### Training Tip: Separate Training From Evaluation

- use `model.train()` before training steps
- use `with torch.no_grad():` when evaluating so PyTorch does not track gradients
- store losses in a list if you want to plot learning progress later


````python
torch.manual_seed(42)

reg_model = nn.Sequential(
    nn.Linear(X_train_tensor.shape[1], 16),
    nn.ReLU(),
    nn.Linear(16, 8),
    nn.ReLU(),
    nn.Linear(8, 1),
)

reg_loss_fn = nn.MSELoss()
reg_optimizer = torch.optim.Adam(reg_model.parameters(), lr=0.01)

reg_train_losses = []
reg_val_losses = []

for epoch in range(300):
    reg_model.train()
    train_preds = reg_model(X_train_tensor)
    train_loss = reg_loss_fn(train_preds, y_train_reg)

    reg_optimizer.zero_grad()
    train_loss.backward()
    reg_optimizer.step()

    reg_model.eval()
    with torch.no_grad():
        val_preds = reg_model(X_val_tensor)
        val_loss = reg_loss_fn(val_preds, y_val_reg)

    reg_train_losses.append(train_loss.item())
    reg_val_losses.append(val_loss.item())

plt.figure(figsize=(8, 4))
plt.plot(reg_train_losses, label="train loss")
plt.plot(reg_val_losses, label="val loss")
plt.xlabel("epoch")
plt.ylabel("MSE loss")
plt.title("Regression training curve")
plt.legend()
plt.show()

reg_model.eval()
with torch.no_grad():
    test_preds_reg = reg_model(X_test_tensor)
    mae = torch.mean(torch.abs(test_preds_reg - y_test_reg)).item()

print("Regression test MAE:", round(mae, 3))
````

## Day 17 - Binary Classification In PyTorch

### Problem type

Classification means predicting a class label.
Here, `passed_test` is either `0` or `1`, so this is binary classification.

### Important loss choice

For binary classification in PyTorch, a strong default is `nn.BCEWithLogitsLoss()`.

Why?

- it expects raw outputs called logits
- it is numerically stable
- you apply `torch.sigmoid` only when you want probabilities

### Prediction rule

- logits -> `sigmoid` -> probabilities
- probabilities >= 0.5 -> class `1`
- otherwise -> class `0`

### Daily exercises

- Train a classifier for `passed_test`.
- Track training and validation loss.
- Compute test accuracy.
- Print the first 10 probabilities and predicted labels.

### Bootcamp exercise folder

- `Week3/day17_pytorch_classification/README.md`
- Best real classification dataset: `data/kaggle/titanic.csv`
- Bonus multiclass dataset: `data/kaggle/iris.csv`


### Classification Tip: Logits Are Not Probabilities

- `BCEWithLogitsLoss()` expects raw model outputs called logits
- apply `torch.sigmoid(...)` only when you want probabilities for inspection or thresholding
- convert probabilities to classes with a threshold like `>= 0.5`


````python
torch.manual_seed(42)

clf_model = nn.Sequential(
    nn.Linear(X_train_tensor.shape[1], 16),
    nn.ReLU(),
    nn.Linear(16, 8),
    nn.ReLU(),
    nn.Linear(8, 1),
)

clf_loss_fn = nn.BCEWithLogitsLoss()
clf_optimizer = torch.optim.Adam(clf_model.parameters(), lr=0.01)

clf_train_losses = []
clf_val_losses = []

for epoch in range(300):
    clf_model.train()
    train_logits = clf_model(X_train_tensor)
    train_loss = clf_loss_fn(train_logits, y_train_clf)

    clf_optimizer.zero_grad()
    train_loss.backward()
    clf_optimizer.step()

    clf_model.eval()
    with torch.no_grad():
        val_logits = clf_model(X_val_tensor)
        val_loss = clf_loss_fn(val_logits, y_val_clf)

    clf_train_losses.append(train_loss.item())
    clf_val_losses.append(val_loss.item())

plt.figure(figsize=(8, 4))
plt.plot(clf_train_losses, label="train loss")
plt.plot(clf_val_losses, label="val loss")
plt.xlabel("epoch")
plt.ylabel("BCE loss")
plt.title("Classification training curve")
plt.legend()
plt.show()

clf_model.eval()
with torch.no_grad():
    test_logits = clf_model(X_test_tensor)
    test_probs = torch.sigmoid(test_logits)
    test_preds = (test_probs >= 0.5).float()
    test_accuracy = (test_preds == y_test_clf).float().mean().item()

print("Classification test accuracy:", round(test_accuracy, 3))
print("First 10 probabilities:", test_probs[:10].squeeze().numpy())
print("First 10 predictions:", test_preds[:10].squeeze().numpy())
````

## Day 18 - Evaluating Models and Debugging Training

### Regression metrics

- MAE: average absolute error
- MSE: average squared error

### Classification metrics

- accuracy
- precision
- recall

### Common beginner mistakes

- forgetting to normalize features
- using the wrong target shape
- mixing regression and classification loss functions
- evaluating on the training set only
- forgetting `optimizer.zero_grad()`

### Daily exercises

- Compute accuracy, precision, and recall manually.
- Compare train loss and validation loss.
- Write 3 reasons why overfitting can happen.


### Debugging Power-Ups

A few small Python habits can save a lot of time when something looks wrong.

- use `enumerate(items, start=1)` when you need both the position and the value in a loop
- use `zip(a, b)` when two sequences should be processed together
- use `assert X.shape[0] == y.shape[0], "Mismatched rows"` to catch bad assumptions early
- avoid variable names like `list`, `dict`, `str`, or `sum` because they hide Python built-ins
- use quick prints like `print(f"{tensor.shape=}")` or `print(f"{column_name!r}")` when checking shapes and labels


````python
with torch.no_grad():
    tp = ((test_preds == 1) & (y_test_clf == 1)).sum().item()
    tn = ((test_preds == 0) & (y_test_clf == 0)).sum().item()
    fp = ((test_preds == 1) & (y_test_clf == 0)).sum().item()
    fn = ((test_preds == 0) & (y_test_clf == 1)).sum().item()

precision = tp / (tp + fp) if (tp + fp) > 0 else 0.0
recall = tp / (tp + fn) if (tp + fn) > 0 else 0.0
accuracy = (tp + tn) / (tp + tn + fp + fn)

print({
    "tp": tp,
    "tn": tn,
    "fp": fp,
    "fn": fn,
    "accuracy": round(accuracy, 3),
    "precision": round(precision, 3),
    "recall": round(recall, 3),
})
````

````python
def prepare_tabular_data(df, feature_columns, target_column, train_frac=0.7, val_frac=0.15, seed=42):
    shuffled = df.sample(frac=1, random_state=seed).reset_index(drop=True)
    n_total = len(shuffled)
    n_train = int(n_total * train_frac)
    n_val = int(n_total * val_frac)

    train = shuffled.iloc[:n_train].copy()
    val = shuffled.iloc[n_train:n_train + n_val].copy()
    test = shuffled.iloc[n_train + n_val:].copy()

    means = train[feature_columns].mean()
    stds = train[feature_columns].std().replace(0, 1)

    X_train = (train[feature_columns] - means) / stds
    X_val = (val[feature_columns] - means) / stds
    X_test = (test[feature_columns] - means) / stds

    y_train = train[[target_column]]
    y_val = val[[target_column]]
    y_test = test[[target_column]]

    return X_train, X_val, X_test, y_train, y_val, y_test


def to_float_tensor(dataframe):
    return torch.tensor(dataframe.values, dtype=torch.float32)


example_features = feature_columns
X_train_df2, X_val_df2, X_test_df2, y_train_df2, y_val_df2, y_test_df2 = prepare_tabular_data(
    engineered_df.fillna(engineered_df.mean(numeric_only=True)),
    example_features,
    "passed_test",
)

print("Prepared pipeline data shapes:")
print(X_train_df2.shape, X_val_df2.shape, X_test_df2.shape)
````
